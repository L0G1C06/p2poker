```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = pd.read_csv("./data/synthetic_poker_game_data_large.csv")
dataset.head()
```

## Distribuição de vitórias por jogador

```python
victory_distribution = dataset['winner'].value_counts()
plt.figure(figsize=(8,5))
sns.barplot(x=victory_distribution.index, y=victory_distribution.values, palette='viridis')
plt.title("Distribuition of Victories by Player")
plt.xlabel("Player")
plt.ylabel("Number of Wins")
plt.show()
```

## Tamamho médio do pote final

```python
average_pot = dataset['total pot bet'].mean()
print(f"Average Total Pot Bet: {average_pot:.2f}")
```
